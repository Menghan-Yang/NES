```python
from pliers.extractors import BrightnessExtractor
from pliers.stimuli import VideoStim
from pliers.filters import FrameSamplingFilter
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import os
import glob
import numpy as np
from tqdm.notebook import tqdm
current_directory = '/Storage/Data/nes_menghan/code_menghan/NeuroExperience/'
svg_file = 'svg_files' ## svg plots
```

```python
def extract_video_brightness(video_path,Samplingrate):
    # Load the video
    video = VideoStim(video_path)
    
    frame_filter = FrameSamplingFilter(hertz=Samplingrate)
    frames = frame_filter.transform(video)

    brightness = BrightnessExtractor()
    brightness_results = brightness.transform(frames)
    
    
    # Combine results into a DataFrame
    brightness_df = pd.concat([r.to_df() for r in brightness_results], ignore_index=True)
    # Extract the brightness values
    brightness_values = brightness_df['brightness'].values
    return brightness_values.tolist()
```

```python
# Example usage
video_path = '/Storage/Data/nes_menghan/liquidsword2_materials/Video_final_scanning/Video_Scan/'  # Replace with your video file path
# brightness_values = extract_video_brightness(video_path)
```

```python
video_brightness_all = []
for video_num in tqdm(range(24)):
    video_lab = video_num+1
    video_nam = 'Scan_Video'+str(video_lab)+'.mp4'
    video_path_name = os.path.join(video_path, video_nam)
    brightness_values = extract_video_brightness(video_path_name,1/2)
    video_brightness_all.append(brightness_values)
    
```

```python
onset_movie = os.path.join('/Storage/Data/nes_menghan/', 'code_menghan', 'NeuroExperience','onset_movie.txt')

onset_movie = pd.read_csv(onset_movie,sep='\t').set_index('Movie name')
onset_movie
```

### create brightness design matrix

```python
video_DM = np.zeros((2966,24))
for video_num in tqdm(range(24)):
    video_DM[onset_movie['Entire Onset'][video_num]:onset_movie['Entire Onset'][video_num]+len(video_brightness_all[video_num]),video_num] = video_brightness_all[video_num]
```

```python
run_intercept = np.zeros((2966,5))
# run_TR = [0,588,613,598,588,579]
run_TR = [0,588,1201,1799,2387,2966]
for video_run in tqdm(range(5)):
    run_intercept[run_TR[video_run]:run_TR[video_run+1],video_run] = 1
```

```python
DM_all = np.hstack([video_DM, run_intercept])
DM_all.shape
```

```python
data_dir = '/Storage/Data/nes_menghan/preprocessed/final/'
subj_sids_list = os.listdir(data_dir)
subj_sids_iteration = []
for subj in subj_sids_list:
    subj_path = os.path.join(data_dir, subj)
    subj_path_sess = os.listdir(subj_path)
    if len(subj_path_sess) == 5:
        subj_sids_iteration.append(subj)
    
    
subj_sids_iteration.remove('sub-sid001215') ###the Timepoints of this subject's third session were fewer than others
subj_sids_iteration.remove('sub-sid001473') ### the fifth

subj_sids_iteration.remove('sub-sid000860') 
subj_sids_iteration.remove('sub-sid001201')  ### head coil issue: HEP
print(len(subj_sids_iteration))
subj_sids_iteration
```

```python
from nltools.stats import regress as regression

session_num = 5
task_num = 5
roi_num = 4
video_spatiotemporal_all = []
b_pd_session = pd.DataFrame()


for sub_name in tqdm(subj_sids_iteration):
    data_subj_session = pd.DataFrame([])
    subj_spatiotemporal = []
    isc_data = glob.glob(os.path.join('/Storage/Data/nes_menghan/denoised/Denoised_ROI_spatial',f"{sub_name}_ses*_task-movie*_Denoised_ROI{roi_num}.csv"))
    isc_data.sort()

    isc_data_test = pd.read_csv(isc_data[0])
    subj_data_spatiotemporal = np.empty((2966,isc_data_test.shape[1],5))

    for i in range(session_num):
        isc_data_session = isc_data[i*5:i*5+5]
    #             print(i,isc_data_session)
        data_subj_session = pd.DataFrame([])
        for j in range(task_num):
            isc_data_session_task = pd.read_csv(isc_data_session[j])
            if len(data_subj_session) == 0:
                data_subj_session = isc_data_session_task
            else:
                data_subj_session = np.concatenate((data_subj_session,isc_data_session_task),axis = 0)

        subj_data_spatiotemporal[:,:,i] = data_subj_session


        for m in range(task_num):
            subj_data_spatiotemporal_session = subj_data_spatiotemporal[:,:,m]
            subj_data_spatiotemporal_session_mean = subj_data_spatiotemporal_session.mean(axis = 1)
            b, se, t, p, df, res = regression(DM_all, subj_data_spatiotemporal_session_mean, mode="ols")

            # b_pd = b.mean(axis = 1)
            b_pd = pd.DataFrame(b[:24],columns=['beta estimation'])
            b_pd['video'] = onset_movie.index
            b_pd['participant'] = sub_name
            b_pd['session'] = m+1

            b_pd_session = pd.concat([b_pd_session,b_pd],axis = 0)



```

## Voxel level activation with Monotonic contrast

```python
from nltools.stats import regress as regression
from nltools.data import Brain_Data, Adjacency
from nltools.data import Design_Matrix

session_num = 5
task_num = 5
excluded_files = glob.glob(os.path.join('/Storage/Data/nes_menghan/denoised/smoothed_unfilter','*_rec-moco_*'))

for sub_name in tqdm(subj_sids_iteration):
# sub_name = subj_sids_iteration[0]
    for sess_num in tqdm(range(5)):
        sess = sess_num+1

        voxel_data = glob.glob(os.path.join('/Storage/Data/nes_menghan/denoised/smoothed_unfilter',f"{sub_name}_ses-{sess}*_task-movie*"))
        filtered_voxel_data = [f for f in voxel_data if f not in excluded_files]
        filtered_voxel_data.sort()

        subj_data = Brain_Data()
        for a in filtered_voxel_data:
            subj_data = subj_data.append(Brain_Data(a))
        
        subj_data.X = Design_Matrix(DM_all)
        stats_data = subj_data.regress()
        
        for video in range(24):
            stats_data['beta'][video].write(os.path.join('/Storage/Data/nes_menghan/results/visual_beta', f'{sub_name}_sess-{sess}_video-{video+1}_beta.nii.gz'))
        
```

```python
for sub_name in tqdm(subj_sids_iteration):
    for video in tqdm(range(24)):
        beta_data = glob.glob(os.path.join('/Storage/Data/nes_menghan/results/visual_beta',f"{sub_name}_sess-*_video-{video+1}_*"))
        beta_data.sort()
        subj_beta_data = Brain_Data(beta_data)
        subj_beta_data_mono = subj_beta_data*[-2,-1,0,1,2]
        subj_beta_data_mono.write(os.path.join('/Storage/Data/nes_menghan/results/visual_beta', f'{sub_name}_video-{video+1}_monotonic.nii.gz'))
```

```python
import numpy as np
import pandas as pd
import nibabel as nib
from nilearn import image
from nilearn.datasets import fetch_atlas_harvard_oxford
from nilearn.reporting import get_clusters_table

# 1) Your (already-thresholded) image
stat_img = beta_data_mono_all.ttest(threshold_dict={'fdr': .05})['thr_t'].to_nifti()

# 2) All clusters, both signs
tbl = get_clusters_table(
    stat_img=stat_img,
    stat_threshold=0.0,   # fine because already FDR-thresholded
    cluster_threshold=0,
    two_sided=True,
    min_distance=8
)

# ---- normalize column names to find what we need ----
def norm(s): 
    return s.lower().replace(' ', '').replace('_', '').replace('-', '')

col_map = {c: norm(c) for c in tbl.columns}

# XYZ columns (your table has 'X','Y','Z')
rev = {v:k for k,v in col_map.items()}
xcol = rev.get('x') or rev.get('peakx')
ycol = rev.get('y') or rev.get('peaky')
zcol = rev.get('z') or rev.get('peakz')

# t/z column (yours is 'Peak Stat')
tcol = (rev.get('peakstat') or rev.get('t') or rev.get('tvalue') or
        rev.get('z') or rev.get('zscore') or rev.get('stat') or rev.get('maxstat'))
if not all([xcol,ycol,zcol,tcol]):
    raise ValueError(f"Could not detect columns. Got: {list(tbl.columns)}")

# 3) Atlas for labeling (Harvard–Oxford cortical)
atlas = fetch_atlas_harvard_oxford('cort-maxprob-thr25-2mm')
atlas_path = getattr(atlas, 'maps', None) or getattr(atlas, 'filename', None)
atlas_img = image.resample_to_img(atlas_path, stat_img, interpolation='nearest')
atlas_data = atlas_img.get_fdata().astype(int)
aff = atlas_img.affine
labels = np.array(getattr(atlas, 'labels', []))

def mni_to_label(xyz_mni):
    ijk = np.round(nib.affines.apply_affine(np.linalg.inv(aff), xyz_mni)).astype(int)
    if np.any(ijk < 0) or np.any(ijk >= atlas_data.shape):
        return 'Out of atlas bounds'
    idx = atlas_data[tuple(ijk)]
    return labels[idx] if idx < len(labels) else 'Unknown'

# 4) Build the final table
out = tbl[[xcol, ycol, zcol, tcol]].copy()
out.rename(columns={tcol: 't_value'}, inplace=True)
out['Region'] = out[[xcol, ycol, zcol]].apply(
    lambda r: mni_to_label([float(r[xcol]), float(r[ycol]), float(r[zcol])]),
    axis=1
)
out = out[[xcol, ycol, zcol, 'Region', 't_value']]

# Show (and optionally save)
print(out.to_string(index=False))
# out.to_csv('cluster_peaks_xyz_region_t.csv', index=False)
```
